```python
from langchain_core.tools import tool
import requests

@tool("get_products", description="Get products from the store filtered by price")
def get_products(price: int ):
    """Get products from the store."""
    response = requests.get("https://api.escuelajs.co/api/v1/products")
    products = response.json()
    return "".join([f"{p['title']} - ${p['price']}" for p in products])
    # products = [
    #     {"name": "Product 1", "price": 10},
    #     {"name": "Product 2", "price": 20},
    #     {"name": "Product 3", "price": 30},
    # ]
    # return "".join([f"{p['name']} - ${p['price']}" for p in products if p["price"] <= price])
```

```python
get_products.invoke({"price": 20})
```
